### Feature Engineering
- Age of card holder
- hour of transactions (fraudsters tend to use cards at odd hours when the card owner may not be active)
- day of week (fraudsters are more likely to use the card during weekends than weekdays)
- distance_km (distance between card holder and mechant)
- card_txn_count_24h (number of transactions on this card in the trailing 24h (velocity).)
- rate - card_amt_mean / card_amt_std — the card's typical spend, so the model can spot a transaction that's abnormal for that card. Same past-only rule.
- amt_vs_card_mean 

```python
# Age
new_train_ds['age'] = (new_train_ds['trans_date_trans_time'] - new_train_ds['dob']).dt.days // 365

new_train_ds.head()
```

```python
# Hour and Day of transaction
new_train_ds['hour'] = new_train_ds['trans_date_trans_time'].dt.hour

new_train_ds['day'] = new_train_ds['trans_date_trans_time'].dt.day

new_train_ds.head()
```

```python
# Distance in kilometers
def haversine_km(lat1, lon1, lat2, lon2):
    R = 6371
    # convert degrees to radians
    lat1, lon1, lat2, lon2 = map(np.radians, [lat1, lon1, lat2, lon2])

    dlat = lat2 - lat1
    dlon = lon2 - lon1

    a = np.sin(dlat / 2)**2 + np.cos(lat1) * np.cos(lat2) * np.sin(dlon / 2)**2
    return R * 2 * np.arcsin(np.sqrt(a))

new_train_ds["distance_km"] = haversine_km(
    new_train_ds["lat"], new_train_ds["long"],
    new_train_ds["merch_lat"], new_train_ds["merch_long"]
)

new_train_ds.head()
```

```python
# card_txn_count_24h
new_train_ds["card_txn_count_24h"] = (
    new_train_ds.set_index("trans_date_trans_time")
      .groupby("cc_num")["amt"]
      .rolling("24h", closed="left")
      .count()
      .values
)

new_train_ds["card_txn_count_24h"] = new_train_ds["card_txn_count_24h"].fillna(0)
```

```python
new_train_ds["card_txn_count_24h"].describe(include='all')
```

```python
# card_amt_mean
new_train_ds["card_amt_mean"] = (
    new_train_ds
    .groupby("cc_num")["amt"]
    .transform(lambda s: s.shift().expanding().mean())
).fillna(new_train_ds['amt'])

```

```python
# card_amt_mean
new_train_ds['card_amt_std'] = (
    new_train_ds
    .groupby('cc_num')['amt']
    .transform(lambda s: s.shift().expanding().std())
).fillna(0)
```

```python
rate = new_train_ds['card_amt_mean'] / (new_train_ds['card_amt_std'])
```

```python
new_train_ds["rate"] = rate.replace([np.inf, -np.inf], np.nan).fillna(0)
```
