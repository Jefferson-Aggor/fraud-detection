### Feature Engineering

- Age of card holder
- hour of transactions (fraudsters tend to use cards at odd hours when the card owner may not be active)
- day of week (fraudsters are more likely to use the card during weekends than weekdays)
- distance_km (distance between card holder and mechant)
- card_txn_count_24h (number of transactions on this card in the trailing 24h (velocity).)
- rate - card_amt_mean / card_amt_std — the card's typical spend, so the model can spot a transaction that's abnormal for that card. Same past-only rule.
- amt_vs_card_mean


```python
# Age
def calc_age(data, col='age'):
    data[col] = (new_train_ds['trans_date_trans_time'] - new_train_ds['dob']).dt.days // 365
    return data


new_train_ds = calc_age(new_train_ds)

new_train_ds.head()
```

```python
test_ds = calc_age(test_ds)

test_ds.head()
```

```python
# Hour and Day of transaction
def calc_hour_and_day(data):
    data['hour'] = data['trans_date_trans_time'].dt.hour
    data['day'] = data['trans_date_trans_time'].dt.day

    return data

new_train_ds = calc_hour_and_day(new_train_ds)

new_train_ds.head()
```

```python
test_ds = calc_hour_and_day(test_ds)

test_ds.head()
```

```python
# Distance in kilometers
def haversine_km(lat1, lon1, lat2, lon2):
    R = 6371
    # convert degrees to radians
    lat1, lon1, lat2, lon2 = map(np.radians, [lat1, lon1, lat2, lon2])

    dlat = lat2 - lat1
    dlon = lon2 - lon1

    a = np.sin(dlat / 2)**2 + np.cos(lat1) * np.cos(lat2) * np.sin(dlon / 2)**2
    return R * 2 * np.arcsin(np.sqrt(a))

new_train_ds["distance_km"] = haversine_km(
    new_train_ds["lat"], new_train_ds["long"],
    new_train_ds["merch_lat"], new_train_ds["merch_long"]
)

new_train_ds.head()
```

```python
test_ds["distance_km"] = haversine_km(
    test_ds["lat"], test_ds["long"],
    test_ds["merch_lat"], test_ds["merch_long"]
)

test_ds.head()
```

```python
# card_txn_count_24h
def calc_card_txn_count_24h(data, col='card_txn_count_24h'):
  data[col] = (
      data.set_index("trans_date_trans_time")
        .groupby("cc_num")["amt"]
        .rolling("24h", closed="left")
        .count()
        .values
  )

  return data

new_train_ds = calc_card_txn_count_24h(new_train_ds).fillna(0)

new_train_ds.head()
```

```python
test_ds = calc_card_txn_count_24h(test_ds).fillna(0)

test_ds.head()
```

```python
# card_amt_mean
def calc_card_amt_mean(data, col='card_amt_mean'):
    data["card_amt_mean"] = (
        data
        .groupby("cc_num")["amt"]
        .transform(lambda s: s.shift().expanding().mean())
    ).fillna(data['amt'])

    return data

new_train_ds = calc_card_amt_mean(new_train_ds)

new_train_ds.head()
```

```python
test_ds = calc_card_amt_mean(test_ds)

test_ds.head()
```

```python
# card_amt_mean
def calc_card_amt_std(data, col='card_amt_std'):
    data[col] = (
        data
        .groupby('cc_num')['amt']
        .transform(lambda s: s.shift().expanding().std())
    ).fillna(0)

    return data

new_train_ds = calc_card_amt_std(new_train_ds)

new_train_ds.head()
```

```python
test_ds = calc_card_amt_std(test_ds)

test_ds.head()
```

```python
def rate(data, col='rate'):
    rate = new_train_ds['card_amt_mean'] / (new_train_ds['card_amt_std'])
    data[col] = rate.replace([np.inf, -np.inf], np.nan).fillna(0)
    
    return data
```

```python
new_train_ds = rate(new_train_ds)

new_train_ds.head()
```

```python
test_ds = rate(test_ds)

test_ds.head()
```

```python
# amt_vs_card_mean

def calc_amt_vs_card_mean(data, col='amt_vs_card_mean'):
    mean = data['card_amt_mean'].replace([np.inf, -np.inf], np.nan).fillna(0)
    data[col] = data['amt'] / mean
    
    return data

new_train_ds = calc_amt_vs_card_mean(new_train_ds)

new_train_ds.head()
```

```python
test_ds = calc_amt_vs_card_mean(test_ds)

test_ds.head()
```

```python
# One Hot Encoding Category
new_train_ds = pd.get_dummies(new_train_ds, columns=['category'], drop_first=True)
test_ds = pd.get_dummies(test_ds, columns=['category'], drop_first=True)
```

```python
# Frequency Encoding
# def frequency_enc(data, col):
#     freq = data[col].value_counts
#     return data[col].map(freq).fillna(0)

# new_train_ds['mechant_enc'] = frequency_enc(new_train_ds, 'merchant')
```

```python
def freq_enc(data, target_col, col_name):
    freq = data[target_col].value_counts()
    data[col_name] = data[target_col].map(freq)

    return data
```

```python
new_train_ds = freq_enc(new_train_ds, 'merchant', 'merchant_enc')
new_train_ds = freq_enc(new_train_ds, 'state', 'state_enc')

new_train_ds.head()
```

```python
test_ds = freq_enc(test_ds, 'merchant', 'merchant_enc')
test_ds = freq_enc(test_ds, 'state', 'state_enc')

test_ds.head()
```
